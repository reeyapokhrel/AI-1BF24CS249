```python
board = [" " for _ in range(9)]

def show():
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} ")

def win(p):
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    return any(all(board[pos] == p for pos in condition) for condition in win_conditions)
def ai():
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            return
while True:
    show()
    move = int(input("Enter position (1-9): ")) - 1
    if move < 0 or move > 8 or board[move] != " ":
        print("Invalid move")
        continue
    board[move] = "X"
    if win("X"):
        show()
        print("You win!")
        break
    ai()
    if win("O"):
        show()
        print("AI wins!")
        break
    if " " not in board:
        show()
        print("Draw!")
        break
```

```python
def vacuum_cleaner():
    room_A = input("Is Room A dirty? (yes/no): ").lower()
    room_B = input("Is Room B dirty? (yes/no): ").lower()
    position = input("Enter vacuum position (A/B): ").upper()
    print("\nInitial State:")
    print("Room A:", room_A)
    print("Room B:", room_B)
    print("Vacuum Position:", position)
    while room_A == "yes" or room_B == "yes":
        if position == "A":
            if room_A == "yes":
                print("\nAction: SUCK")
                room_A = "no"
                print("Room A cleaned.")
            elif room_B == "yes":
                print("\nAction: RIGHT")
                position = "B"
                print("Vacuum moved to Room B.")
        elif position == "B":
            if room_B == "yes":
                print("\nAction: SUCK")
                room_B = "no"
                print("Room B cleaned.")
            elif room_A == "yes":
                print("\nAction: LEFT")
                position = "A"
                print("Vacuum moved to Room A.")
    print("\nFinal State:")
    print("Room A: Clean")
    print("Room B: Clean")
    print("Vacuum Position:", position)
    print("\nGoal achieved! Both rooms are clean.")


vacuum_cleaner()
```
